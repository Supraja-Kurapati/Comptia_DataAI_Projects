```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk import ngrams
from collections import Counter
```

```python
sentence="The quick brown fox jumps over the lazy dog"
tokens= nltk.word_tokenize(sentence.lower())
def generate_ngrams(tokens,n):
    return list (ngrams(tokens,n))
unigrams=generate_ngrams(tokens,1)
bigrams=generate_ngrams(tokens,2)
trigrams=generate_ngrams(tokens,3)
print("Unigrams:")
print(unigrams)
print("\nBigrams:")
print(bigrams)
print("\nTrigrams:")
print(trigrams)
```

```python
# count freq of n-grams
bigram_freq=Counter(bigrams)
print("\nBigram Frequencies:")
print(bigram_freq)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score
import pandas as pd

# Define a real world sample news dataset
news_data= [
    # Sports
    {"text": "The striker scored a stunning hat-trick in the final minutes of the match to secure the championship victory.", "category": "Sports"},
    {"text": "Olympic athletes are arriving at the village ahead of the opening ceremony and track events.", "category": "Sports"},
    {"text": "The basketball coach signed a record-breaking contract extension after leading the team to the playoffs.", "category": "Sports"},

    # Politics
    {"text": "The Prime Minister announced a new legislative bill targeting tax reform and government spending cuts.", "category": "Politics"},
    {"text": "Voters are heading to the polling stations today for the high-stakes parliamentary election.", "category": "Politics"},
    {"text": "The senate debated the new foreign policy treaty for over six hours before calling a vote.", "category": "Politics"},

    # Business
    {"text": "Shares of the tech giant surged by 8% following a massive increase in quarterly revenue and profit margins.", "category": "Business"},
    {"text": "The central bank raised interest rates by a quarter percentage point to combat rising inflation.", "category": "Business"},
    {"text": "Two major automotive companies announced a multi-billion dollar merger to accelerate electric vehicle production.", "category": "Business"}
]

# Convert to a DataFrame for easy manipulation
df = pd.DataFrame(news_data)

# 2. Split data into features (X) and labels (y), then into Train/Test sets
X = df['text']
y = df['category']

# Using a 2/3 split for this tiny dataset so every class is represented in the test set
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42, stratify=y)

# 3. Vectorize text using TF-IDF
# sublinear_tf=True scales down the impact of words repeated many times in a single long article
tfidf_vectorizer = TfidfVectorizer(stop_words='english', sublinear_tf=True)

X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)

# 4. Train a Logistic Regression Classifier
# Logistic Regression is highly effective and fast for high-dimensional TF-IDF vectors
classifier = LogisticRegression(random_state=42)
classifier.fit(X_train_tfidf, y_train)

# 5. Evaluate the model
predictions = classifier.predict(X_test_tfidf)
print(f"Model Accuracy: {accuracy_score(y_test, predictions) * 100:.2f}%\n")
print("Classification Report:")
print(classification_report(y_test, predictions))

# 6. Test with unseen, simulated real-time breaking news headlines
unseen_headlines = [
    "Farhan Akhtar's directorial Dil Chahta Hai recently completed 25 years, ",
    " and the film's cast and director came together to recall their time shooting the cult classic. In a recent interaction, Aamir Khan ",
    "and Farhan Akhtar recalled shooting a crucial scene in the film in which Akshaye Khanna slapped Aamir almost 17 times."
]

# Transform the fresh text to the existing TF-IDF structure
unseen_tfidf = tfidf_vectorizer.transform(unseen_headlines)
predicted_categories = classifier.predict(unseen_tfidf)

print("\n--- Live Classification Results ---")
for headline, category in zip(unseen_headlines, predicted_categories):
    print(f"Headline: '{headline}'\n--> Predicted Category: [{category}]\n")
```

```python
import sys
!{sys.executable} -m pip install --upgrade --force-reinstall networkx node2vec matplotlib gensim "numpy<2.0.0" "scipy==1.10.1"

import networkx as nx
from node2vec import Node2Vec
import matplotlib.pyplot as plt


G = nx.Graph()


friendships = [
    ("Alice", "Bob"), ("Alice", "Charlie"), ("Bob", "Charlie"), # Finance Team Cluster
    ("David", "Eva"), ("David", "Frank"), ("Eva", "Frank"),    # Engineering Team Cluster
    ("Charlie", "David") # The "Bridge" connection between the two separate departments
]
G.add_edges_from(friendships)


node2vec = Node2Vec(G, dimensions=16, walk_length=5, num_walks=100, workers=1)

# 3. Train the model to generate the embeddings
model = node2vec.fit(window=3, min_count=1, batch_words=4)

# 4. View and use the embeddings
print("--- Generated Node Embeddings ---")
for node in G.nodes():
    vector = model.wv[node]
    # Truncating output for readability
    print(f"Node: {node:8} | Vector Snippet: {vector[:4]}...")

print("\n--- Measuring Network Similarity ---")
# Because Charlie and David are connected directly, they should be relatively close
sim_connected = model.wv.similarity("Charlie", "David")
print(f"Similarity between Bridge Nodes (Charlie & David): {sim_connected:.4f}")

# Alice and Frank are completely isolated on opposite sides of the network
sim_isolated = model.wv.similarity("Alice", "Frank")
print(f"Similarity between Distant Nodes (Alice & Frank):   {sim_isolated:.4f}")

# 5. Find the most structurally similar individuals
print("\nMost similar individuals to 'Alice':")
for friend, similarity in model.wv.most_similar("Alice")[:2]:
    print(f"- {friend}: {similarity:.4f}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import make_pipeline
from lime.lime_text import LimeTextExplainer

# 1. Dummy NLP Training Data (Sentiment Analysis)
train_texts = [
    "I absolutely loved this movie, it was incredible!",
    "Best film of the year, highly recommend.",
    "Horrible acting and a terrible plot. Waste of time.",
    "It was okay, but pretty boring and slow.",
    "Fantastic performance by the lead actor, truly spectacular.",
    "Disastrous choice of director. The script was completely garbage."
]
# 1 = Positive, 0 = Negative
train_labels = [1, 1, 0, 0, 1, 0]

# 2. Build a Text Classification Pipeline
# Model-agnostic tools like LIME work cleanly with end-to-end processing pipelines
vectorizer = TfidfVectorizer()
classifier = RandomForestClassifier(n_estimators=100, random_state=42)

pipeline = make_pipeline(vectorizer, classifier)
pipeline.fit(train_texts, train_labels)

# 3. Instantiate the LIME Text Explainer
class_names = ['Negative', 'Positive']
explainer = LimeTextExplainer(class_names=class_names)

# 4. Text instance we want to diagnose
text_to_explain = "The movie was fantastic, though the ending felt a bit terrible."

# 5. Generate Explanation
# Note: pipeline.predict_proba takes raw text strings and outputs class probabilities
exp = explainer.explain_instance(
    text_instance=text_to_explain,
    classifier_fn=pipeline.predict_proba, # Changed predict_fn to classifier_fn
    num_features=4
)

# 6. View the breakdown
print(f"Document to explain: '{text_to_explain}'\n")
print(f"Predicted Class Probabilities:")
probs = pipeline.predict_proba([text_to_explain])[0]
for idx, prob in enumerate(probs):
    print(f"  {class_names[idx]}: {prob:.2%}")
print("-" * 50)

print("Top word contributions for this prediction:")
for word, weight in exp.as_list():
    direction = "Triggers 'Positive'" if weight > 0 else "Triggers 'Negative'"
    print(f"   Word: '{word:<10}' | Weight: {weight:+.4f} | ({direction})")
```

```python
import shap
from transformers import AutoModelForSequenceClassification, AutoTokenizer, pipeline

# 1. Load a pre-trained NLP sentiment analysis model and tokenizer
# We are using a standard DistilBERT model fine-tuned for sentiment analysis
model_name = "distilbert-base-uncased-finetuned-sst-2-english"
model = AutoModelForSequenceClassification.from_pretrained(model_name)
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Create a standard Hugging Face pipeline
# Setting top_k=None ensures the pipeline returns probabilities for all classes
nlp_pipeline = pipeline(
    "sentiment-analysis", model=model, tokenizer=tokenizer, top_k=None
)

# 2. Define the sentences you want to analyze
text_samples = [
    "I absolutely loved the movie, the acting was spectacular and brilliant!",
    "The plot was incredibly boring and a complete waste of time.",
]

# 3. Initialize the SHAP Explainer for NLP
# SHAP automatically detects the pipeline and uses a text-specific partition algorithm
explainer = shap.Explainer(nlp_pipeline)

# 4. Compute SHAP values for the text samples
shap_values = explainer(text_samples)

# 5. Visualize the explanations
# Explain the positive sentiment text
print("Visualizing text explanation for Sample 1...")
# We select index [0, :, 'POSITIVE'] to look at the 'POSITIVE' class output for the first text
shap.plots.text(shap_values[0, :, "POSITIVE"])

# Explain the negative sentiment text
print("Visualizing text explanation for Sample 2...")
shap.plots.text(shap_values[1, :, "NEGATIVE"])
```

```python
import re
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

text = input("Enter any text: ")

# Clean text and split into words
cleaned_text = re.sub(r'[^\w\s]', '', text)
words = cleaned_text.split()

# Filter out stop words
filtered_words = [word for word in words if word.lower() not in stop_words]

print("Filtered Words:", filtered_words)
```
