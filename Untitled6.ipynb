```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense, Embedding
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import to_categorical
```

```python
# 1. Sample textual dataset
text_data = "the quick brown fox jumps over the lazy dog and runs away"
words = text_data.split()
```

```python
# 2. Tokenize words to integer IDs
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text_data])
vocab_size = len(tokenizer.word_index) + 1  # +1 for 0-padding
encoded_words = tokenizer.texts_to_sequences([text_data])[0]
```

```python
# 3. Create sliding window inputs (X) and targets (y)
window_size = 3
X, y = [], []

for i in range(len(encoded_words) - window_size):
    X.append(encoded_words[i:i + window_size])
    y.append(encoded_words[i + window_size])

X = np.array(X)
y = to_categorical(y, num_classes=vocab_size)  # One-hot encode targets
```

```python
# 4. Build the RNN model using an Embedding layer
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=10, input_length=window_size),
    SimpleRNN(50, activation='relu'),
    Dense(vocab_size, activation='softmax')
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
# 5. Train the model
model.fit(X, y, epochs=300, verbose=0)
print("Training complete.")
```

```python
# 6. Predict the next word
test_phrase = "The"
test_sequence = tokenizer.texts_to_sequences([test_phrase])[0]
test_input = np.array([test_sequence])  # Shape: (1, 3)

predicted_probs = model.predict(test_input, verbose=0)
predicted_id = np.argmax(predicted_probs, axis=-1)[0]

# Convert predicted index back to word
predicted_word = tokenizer.index_word.get(predicted_id, "Unknown")
print(f"Input phrase: '{test_phrase}'")
print(f"Predicted next word: '{predicted_word}'")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# 1. Sample Text & ASCII Encoding
text = "hello world! deep learning with character prediction."

# Convert characters to their ASCII integer representations (0-127)
ascii_sequence = [ord(char) for char in text]

# Prepare sequence dataset (sliding window)
seq_length = 5
inputs = []
targets = []

for i in range(len(ascii_sequence) - seq_length):
    inputs.append(ascii_sequence[i : i + seq_length])
    targets.append(ascii_sequence[i + seq_length])

# Convert to PyTorch Tensors
X = torch.tensor(inputs, dtype=torch.long)  # Shape: (batch_size, seq_length)
y = torch.tensor(targets, dtype=torch.long)  # Shape: (batch_size,)


# 2. Define the RNN Model
class CharRNN(nn.Module):
    def __init__(self, vocab_size=128, embed_dim=16, hidden_dim=64):
        super(CharRNN, self).__init__()
        # Embedding layer maps ASCII integer IDs (0-127) to dense vectors
        self.embedding = nn.Embedding(vocab_size, embed_dim)
        # Simple RNN Layer
        self.rnn = nn.RNN(embed_dim, hidden_dim, batch_first=True)
        # Linear layer mapping hidden states to probabilities across ASCII space (128)
        self.fc = nn.Linear(hidden_dim, vocab_size)

    def forward(self, x):
        out = self.embedding(x)  # (batch_size, seq_length, embed_dim)
        out, _ = self.rnn(out)  # (batch_size, seq_length, hidden_dim)
        out = self.fc(
            out[:, -1, :]
        )  # Take output from the last time-step -> (batch_size, vocab_size)
        return out


model = CharRNN(vocab_size=128, embed_dim=16, hidden_dim=64)

# 3. Loss function and Optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.005)

# 4. Train Model
epochs = 300
for epoch in range(epochs):
    model.train()
    optimizer.zero_grad()

    outputs = model(X)
    loss = criterion(outputs, y)

    loss.backward()
    optimizer.step()

    if (epoch + 1) % 50 == 0:
        print(f"Epoch [{epoch + 1}/{epochs}], Loss: {loss.item():.4f}")

# 5. Predict the Next Character
model.eval()
test_str = "a"
test_ascii = [ord(c) for c in test_str[-seq_length:]]
test_tensor = torch.tensor([test_ascii], dtype=torch.long)

with torch.no_grad():
    prediction_logits = model(test_tensor)
    predicted_ascii = torch.argmax(prediction_logits, dim=-1).item()
    predicted_char = chr(predicted_ascii)

print(f"\nInput phrase: '{test_str}'")
print(f"Predicted next character: '{predicted_char}' (ASCII: {predicted_ascii})")
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import to_categorical

# 1. Recreate Dataset with Proper Spacing
subjects = [
    "python", "java", "javascript", "c++", "sql",
    "html", "css", "rust", "go", "kotlin"
]

corpus = []
for subject in subjects:
    line = f"this is this is this is {subject}"
    corpus.extend([line] * 10)

# 2. Tokenize Text
tokenizer = Tokenizer()
tokenizer.fit_on_texts(corpus)
vocab_size = len(tokenizer.word_index) + 1  # +1 for 0-padding

print("Vocabulary:", tokenizer.word_index)

# 3. Create Fixed Sliding Window Inputs (X) and Targets (y)
window_size = 6
X, y = [], []

for text in corpus:
    token_sequence = tokenizer.texts_to_sequences([text])[0]
    for i in range(len(token_sequence) - window_size):
        X.append(token_sequence[i : i + window_size])
        y.append(token_sequence[i + window_size])

X = np.array(X)
y_categorical = to_categorical(y, num_classes=vocab_size)

print(f"\nInput shape (X): {X.shape}")  # (Samples, window_size)
print(f"Target shape (y): {y_categorical.shape}")

# 4. Build RNN Model with Fixed Window Input Length
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=16),
    SimpleRNN(64, activation='relu'),
    Dense(vocab_size, activation='softmax')
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.summary()

# 5. Train Model
model.fit(X, y_categorical, epochs=100, batch_size=16, verbose=0)

# 6. Predict Next Word Using Fixed Window
def predict_next_word(seed_text):
    tokens = tokenizer.texts_to_sequences([seed_text])[0]

    # Enforce exact window_size input
    if len(tokens) < window_size:
        # Left-pad with zeros if input prompt is shorter than window_size
        tokens = [0] * (window_size - len(tokens)) + tokens
    else:
        # Take the last `window_size` tokens if prompt is longer
        tokens = tokens[-window_size:]

    input_array = np.array([tokens])
    predicted_probs = model.predict(input_array, verbose=0)[0]
    predicted_id = np.argmax(predicted_probs)

    return tokenizer.index_word.get(predicted_id, "Unknown")

# 7. Test Predictions
test_prompts = [
    "this",
    "this is",
    "this is this",
    "this is this is"
]

print("\n--- Model Predictions ---")
for prompt in test_prompts:
    predicted_word = predict_next_word(prompt)
    print(f"Prompt: '{prompt}'  --->  Next Word Prediction: '{predicted_word}'")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader

# 1. Sample Dataset & Tokenization
text = (
    "the quick brown fox jumps over the lazy dog and runs away into the forest "
    "deep learning with recurrent neural networks allows sequence modeling"
)
words = text.lower().split()

# Create Vocabulary
vocab = sorted(list(set(words)))
word2idx = {w: i + 1 for i, w in enumerate(vocab)}  # 0 reserved for padding
idx2word = {i: w for w, i in word2idx.items()}
vocab_size = len(word2idx) + 1

# 2. Dataset Construction (Extracting Window, Prefix, and Suffix)
def create_dataset(words, window_size=3):
    X, y_prefix, y_suffix = [], [], []
    # Loop needs room for 1 prefix word before and 1 suffix word after the window
    for i in range(1, len(words) - window_size):
        prefix_target = word2idx[words[i - 1]]
        window_input = [word2idx[w] for w in words[i : i + window_size]]
        suffix_target = word2idx[words[i + window_size]]

        X.append(window_input)
        y_prefix.append(prefix_target)
        y_suffix.append(suffix_target)

    return (
        torch.tensor(X, dtype=torch.long),
        torch.tensor(y_prefix, dtype=torch.long),
        torch.tensor(y_suffix, dtype=torch.long),
    )

window_size = 5
X, y_prefix, y_suffix = create_dataset(words, window_size=window_size)

# 3. Model Definition with Dual Output Heads
class DualPredictorRNN(nn.Module):
    def __init__(self, vocab_size, embed_dim=32, hidden_dim=64):
        super(DualPredictorRNN, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        self.rnn = nn.RNN(embed_dim, hidden_dim, batch_first=True)

        # Head 1: Predicts the word BEFORE the sequence
        self.prefix_head = nn.Linear(hidden_dim, vocab_size)
        # Head 2: Predicts the word AFTER the sequence
        self.suffix_head = nn.Linear(hidden_dim, vocab_size)

    def forward(self, x):
        # x shape: (batch_size, sequence_length)
        embedded = self.embedding(x)  # (batch_size, sequence_length, embed_dim)
        out, h_n = self.rnn(embedded)

        # Use the first hidden state output for prefix prediction
        prefix_out = self.prefix_head(out[:, 0, :])
        # Use the last hidden state output for suffix prediction
        suffix_out = self.suffix_head(out[:, -1, :])

        return prefix_out, suffix_out

model = DualPredictorRNN(vocab_size=vocab_size, embed_dim=32, hidden_dim=64)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

# 4. Training Loop
epochs = 200
for epoch in range(epochs):
    model.train()
    optimizer.zero_grad()

    prefix_preds, suffix_preds = model(X)

    # Combined Loss for both tasks
    loss_prefix = criterion(prefix_preds, y_prefix)
    loss_suffix = criterion(suffix_preds, y_suffix)
    total_loss = loss_prefix + loss_suffix

    total_loss.backward()
    optimizer.step()

    if (epoch + 1) % 50 == 0:
        print(f"Epoch [{epoch+1}/{epochs}] | Loss: {total_loss.item():.4f}")

# 5. Prediction Function
def predict_context(phrase):
    model.eval()
    tokens = phrase.lower().split()

    # Enforce window size
    if len(tokens) < window_size:
        tokens = ["<pad>"] * (window_size - len(tokens)) + tokens
    else:
        tokens = tokens[-window_size:]

    input_indices = [word2idx.get(w, 0) for w in tokens]
    input_tensor = torch.tensor([input_indices], dtype=torch.long)

    with torch.no_grad():
        prefix_logits, suffix_logits = model(input_tensor)

        pred_prefix_id = torch.argmax(prefix_logits, dim=-1).item()
        pred_suffix_id = torch.argmax(suffix_logits, dim=-1).item()

    pred_prefix = idx2word.get(pred_prefix_id, "<UNK>")
    pred_suffix = idx2word.get(pred_suffix_id, "<UNK>")

    print(f"\nWindow Input:  '{' '.join(tokens)}'")
    print(f"Predicted Prefix:  '{pred_prefix}'")
    print(f"Predicted Suffix:  '{pred_suffix}'")

# Test the model
predict_context("lazy dog and")
predict_context("recurrent neural networks")
```
