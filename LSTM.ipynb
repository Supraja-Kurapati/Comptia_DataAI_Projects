```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

# Training data
X = np.array([
    [28, 29, 30],
    [29, 30, 31],
    [30, 31, 32],
    [31, 32, 33],
    [32, 33, 34]
])
y = np.array([31, 32, 33, 34, 35])

# Convert to 3D: samples, time steps, features
X = X.reshape((X.shape[0], X.shape[1], 1))

model = Sequential([
    LSTM(50, input_shape=(3, 1)),
    Dense(1)
])
model.compile(optimizer="adam", loss="mse")
model.summary()

model.fit(X, y, epochs=100, verbose=0)

new_data = np.array([35, 36, 37]).reshape((1, 3, 1))
prediction = model.predict(new_data, verbose=0)
print("Predicted next value:", prediction[0][0])
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from sklearn.preprocessing import MinMaxScaler

# 1. Generate more sequential data (e.g., sequence length = 3)
# Pattern: y is always the next value after a sequence of 3 consecutive integers
sequence_length = 3
total_numbers = 100

data = np.arange(1, total_numbers + 1)

X_data = []
y_data = []

for i in range(len(data) - sequence_length):
    X_data.append(data[i : i + sequence_length])
    y_data.append(data[i + sequence_length])

X = np.array(X_data)
y = np.array(y_data)

# 2. Scale features (LSTMs perform significantly better with values between 0 and 1)
scaler_X = MinMaxScaler()
scaler_y = MinMaxScaler()

X_scaled = scaler_X.fit_transform(X)
y_scaled = scaler_y.fit_transform(y.reshape(-1, 1))

# 3. Reshape input to 3D [samples, time_steps, features]
X_scaled = X_scaled.reshape((X_scaled.shape[0], sequence_length, 1))

# 4. Define the model matching input_shape to sequence_length (3)
model = Sequential([
    LSTM(64, return_sequences=True, input_shape=(sequence_length, 1)),
    LSTM(32),
    Dense(1)
])

model.compile(optimizer="adam", loss="mse")
model.fit(X_scaled, y_scaled, epochs=200, batch_size=8, verbose=0)

# 5. Predict on new unseen sequence
new_data = np.array([[35, 36, 37]])
new_data_scaled = scaler_X.transform(new_data).reshape((1, sequence_length, 1))

predicted_scaled = model.predict(new_data_scaled, verbose=0)
predicted_value = scaler_y.inverse_transform(predicted_scaled)

print("Predicted next value:", predicted_value[0][0])
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from sklearn.preprocessing import MinMaxScaler

# 1. Dataset Generation (Sequence length of 5)
# Input X: 5 consecutive numbers (e.g. [10, 11, 12, 13, 14])
# Target y: Next number for EACH input step (e.g. [11, 12, 13, 14, 15])
sequence_length = 5
total_numbers = 100
data = np.arange(1, total_numbers + 1)

X_list, y_list = [], []
for i in range(len(data) - sequence_length):
    X_list.append(data[i : i + sequence_length])
    y_list.append(data[i + 1 : i + sequence_length + 1])

X = np.array(X_list)
y = np.array(y_list)

# 2. Scale features to [0, 1] range
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
y_scaled = scaler.transform(y)

# 3. Reshape for LSTM: (samples, time_steps, features) -> (95, 5, 1)
X_scaled = X_scaled.reshape((X_scaled.shape[0], sequence_length, 1))
y_scaled = y_scaled.reshape((y_scaled.shape[0], sequence_length, 1))

# 4. Model Architecture (Matching return_sequences=True with 5 time steps)
model = Sequential([
    LSTM(64, return_sequences=True, input_shape=(sequence_length, 1)),
    Dense(1)
])

model.compile(optimizer="adam", loss="mse")
model.summary()

# 5. Train Model
model.fit(X_scaled, y_scaled, epochs=300, batch_size=4, verbose=0)

# 6. Predict on new sequence [35, 36, 37, 38, 39] -> Expected output: [36, 37, 38, 39, 40]
new_data = np.array([[35, 36, 37, 38, 39]])
new_data_scaled = scaler.transform(new_data).reshape((1, sequence_length, 1))

prediction_scaled = model.predict(new_data_scaled, verbose=0)
prediction = scaler.inverse_transform(prediction_scaled.reshape(1, -1))

print("Input sequence:    ", new_data[0])
print("Predicted sequence:", prediction[0])
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, RepeatVector, TimeDistributed, Dense
from sklearn.preprocessing import MinMaxScaler

# 1. Generate Synthetic Sequence Data
# Input X: 20 consecutive numbers (e.g. [1..20], [2..21], ...)
# Target y: The subsequent 20 numbers (e.g. [21..40], [22..41], ...)
timesteps = 20
total_numbers = 200
raw_data = np.arange(1, total_numbers + 1)

X_list, y_list = [], []
for i in range(len(raw_data) - 2 * timesteps + 1):
    X_list.append(raw_data[i : i + timesteps])
    y_list.append(raw_data[i + timesteps : i + 2 * timesteps])

X = np.array(X_list)
y = np.array(y_list)

# 2. Scale features to [0, 1] for stable LSTM training
scaler = MinMaxScaler()
# Fit scaler on full range of values present in the data
scaler.fit(raw_data.reshape(-1, 1))

X_scaled = scaler.transform(X.reshape(-1, 1)).reshape(X.shape[0], timesteps, 1)
y_scaled = scaler.transform(y.reshape(-1, 1)).reshape(y.shape[0], timesteps, 1)

# 3. Define Encoder-Decoder LSTM Model matching input_shape=(20, 1)
model = Sequential([
    # Encoder
    LSTM(64, input_shape=(timesteps, 1)),
    # Convert one encoded vector back to 20 time steps
    RepeatVector(timesteps),
    # Decoder
    LSTM(64, return_sequences=True),
    # Produce one value at each time step
    TimeDistributed(Dense(1))
])

model.compile(optimizer="adam", loss="mse")
model.summary()

# 4. Train the Model
model.fit(X_scaled, y_scaled, epochs=300, batch_size=8, verbose=0)

# 5. Predict on unseen sequence: [100, 101, ..., 119] -> Expected: [120, 121, ..., 139]
test_seq = np.arange(100, 100 + timesteps)
test_scaled = scaler.transform(test_seq.reshape(-1, 1)).reshape(1, timesteps, 1)

predicted_scaled = model.predict(test_scaled, verbose=0)
predicted_sequence = scaler.inverse_transform(predicted_scaled.reshape(timesteps, 1)).flatten()

print("Input sequence (20 steps):")
print(test_seq)
print("\nPredicted sequence (next 20 steps):")
print(np.round(predicted_sequence, 2))
```

LSTM WITH DATASET

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout, Bidirectional, RepeatVector, TimeDistributed

# Load IMDB dataset
max_features = 5000  # Top 5,000 words
max_len = 100        # Input sequence length (100 tokens)

(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=max_features)

# Pad sequences to ensure uniform shape (samples, 100)
X_train = pad_sequences(X_train, maxlen=max_len)
X_test = pad_sequences(X_test, maxlen=max_len)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

```python
import numpy as np
from tensorflow.keras.datasets import imdb

# 1. Print the shape and first 5 tokenized sequences (Data "Head")
print("X_train Shape:", X_train.shape)
print("\n--- X_train Head (First 5 Samples) ---")
print(X_train[:5])

print("\n--- y_train Head (First 5 Sentiment Labels) ---")
print(y_train[:5])

# 2. Decode the first review (Sample 0) back into readable text
word_index = imdb.get_word_index()

# Shift index by 3 to account for Keras special tokens
reverse_word_index = {value + 3: key for key, value in word_index.items()}
reverse_word_index[0] = "<PAD>"
reverse_word_index[1] = "<START>"
reverse_word_index[2] = "<UNK>"

decoded_review = " ".join([reverse_word_index.get(i, "?") for i in X_train[0]])

print("\n--- Decoded Review Sample 0 ---")
print(decoded_review)
print("\nSentiment:", "Positive (1)" if y_train[0] == 1 else "Negative (0)")
```

BASic LSTM

```python
model_basic = Sequential([
    Embedding(input_dim=max_features, output_dim=64),
    LSTM(64),
    Dense(1, activation='sigmoid')
])

model_basic.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model_basic.summary()
model_basic.fit(X_train, y_train, epochs=3, batch_size=64, validation_data=(X_test, y_test))

# Predict sentiment probability for the first 5 test reviews
probs = model_basic.predict(X_test[:5])
classes = (probs > 0.5).astype(int).flatten()

for i, (prob, cls) in enumerate(zip(probs, classes)):
    label = "Positive" if cls == 1 else "Negative"
    print(f"Review {i+1}: Probability = {prob[0]:.4f} -> Prediction: {label}")
```

Stacked

```python
model_stacked = Sequential([
    Embedding(input_dim=max_features, output_dim=64),
    LSTM(64, return_sequences=True),
    LSTM(32),
    Dense(1, activation='sigmoid')
])

model_stacked.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model_stacked.summary()
model_stacked.fit(X_train, y_train, epochs=3, batch_size=64, validation_data=(X_test, y_test))
# Predict using stacked model
probs = model_stacked.predict(X_test[:5])
classes = (probs > 0.5).astype(int).flatten()

for i, (prob, cls) in enumerate(zip(probs, classes)):
    label = "Positive" if cls == 1 else "Negative"
    print(f"Review {i+1}: Probability = {prob[0]:.4f} -> Prediction: {label}")
```

Multivariate

```python
from tensorflow.keras.layers import Input, Concatenate
from tensorflow.keras.models import Model

# Feature 1: Text sequence tokens
text_input = Input(shape=(max_len,), name="text_input")
x = Embedding(input_dim=max_features, output_dim=64)(text_input)
x = LSTM(64)(x)

# Feature 2: Additional numeric feature (e.g., Normalized word count of reviews)
meta_input = Input(shape=(1,), name="meta_input")

# Combine both inputs
combined = Concatenate()([x, meta_input])
output = Dense(1, activation='sigmoid')(combined)

model_multi = Model(inputs=[text_input, meta_input], outputs=output)
model_multi.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model_multi.summary()

# Dummy metadata feature (e.g., sequence length normalized)
review_lengths_train = np.array([len(seq) for seq in X_train]) / max_len
review_lengths_test = np.array([len(seq) for seq in X_test]) / max_len

model_multi.fit(
    [X_train, review_lengths_train], y_train,
    epochs=3, batch_size=64,
    validation_data=([X_test, review_lengths_test], y_test)
)
# Predict using text sequences + review length feature
probs = model_multi.predict([X_test[:5], review_lengths_test[:5]])
classes = (probs > 0.5).astype(int).flatten()

for i, (prob, cls) in enumerate(zip(probs, classes)):
    label = "Positive" if cls == 1 else "Negative"
    print(f"Sample {i+1}: Probability = {prob[0]:.4f} -> Prediction: {label}")
```
